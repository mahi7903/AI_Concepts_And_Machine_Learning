```python
dataset = dataset.drop("City", axis=1)#dropping the columns which either has string or is of no use 
```

```python
dataset = dataset.drop("Sleep Duration", axis=1)
```

```python
dataset = dataset.drop("Have you ever had suicidal thoughts ?", axis=1)
```

```python
dataset = dataset.drop("Degree", axis=1)
```

```python
dataset = dataset.drop("Gender", axis=1)
```

```python
dataset = dataset.drop("id", axis=1)
```

```python
dataset = dataset.drop("Study Satisfaction", axis=1)
```

```python
dataset = dataset.drop("Job Satisfaction", axis=1)
```

```python
Family_History_of_Mental_Illness = pd.get_dummies(dataset["Family History of Mental Illness"])  
dataset = dataset.drop("Family History of Mental Illness", axis=1)  
dataset = pd.concat([dataset, Family_History_of_Mental_Illness], axis=1)  #making the dummies to change the string data into 0's and 1's like true or false 
```

```python
dataset#checking has it been done properly
```

```python
X = dataset[['No','Yes','Age','CGPA','Financial Stress','Work/Study Hours','Academic Pressure','Work Pressure']]  
Y = dataset['Depression']
dataset#alocating x and y columns mannually as it was giving error like not defined also wanted to set the target column 
```

```python
X_train, X_test, y_train, Y_test = train_test_split(X, Y, test_size=0.1)#training the model
```

```python
#algorithm 1
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(max_iter=500, C=1, random_state=10)#increased the itertion from default 100 and c set to 1 as it gives low performance when set to 0.1 or lower
classifier.fit(X_train, y_train)#compiling the model
predictions = classifier.predict(X_test)
print(predictions)

print("Predictions ar:", predictions)#printing out the predictions and the accuracy of the model
accuracy = accuracy_score(Y_test, predictions)
print("Accuracy of the model:", accuracy)
```

```python
#algorithm 2
from sklearn.svm import SVC
classifier = SVC(kernel='linear', C=0.1, random_state=10)# using linear the graph is linear also, lower value of c to improve performance as it is not sam eon 0.5,100,200,5,10,1000 its lower then now 
classifier.fit(X_train, y_train)
predictions = classifier.predict(X_test)
print(predictions)

print("Predictions ar:", predictions)
accuracy = accuracy_score(Y_test, predictions)
print("Accuracy of the model:", accuracy)
```

```python
print("Confusion Matrix:\n", confusion_matrix(Y_test, predictions)) #printing out hte confusion matric
print("Classification Report:\n", classification_report(Y_test, predictions))#printing out the classification report as shown in the lecture
```
