```python
#regression NN model
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import RMSprop #importing the library with different optimizer found out that this is better for regression 
```

```python
dataset = pd.read_csv("Test_Anemia.csv")
```

```python
X = dataset[['GENDER', 'IDENTIFICATION', 'MCH', 'MCHC', 'MCV']]  # alocatting the columns 
y = dataset['Hemoglobin']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=10)
```

```python
model = Sequential([
    Dense(10, activation='relu', input_shape=(X_train.shape[1],)),  
    Dense(10, activation='relu'),  
    Dense(10, activation='relu'),  
    Dense(1, activation='linear')  #output layer last one with linear activation not the softmax as per lecture just to try something new
])
```

```python
model.compile(optimizer=RMSprop(learning_rate=0.001), loss='mean_squared_error', metrics=['mae']) #compiling the model by giving some parametrs
```

```python
model_metrics = model.fit(X_train, y_train, epochs=50, batch_size=16, validation_split=0.1, verbose=1) #training the model for prediction
```

```python
predictions = model.predict(X_test) #prediting the model with the test column
print(predictions)
```

```python
from sklearn.metrics import r2_score #printing the accuracy
r2 = r2_score(y_test, predictions)
print('R2 is:', r2)
```
