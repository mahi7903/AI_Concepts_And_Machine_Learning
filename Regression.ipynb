```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn import metrics
#importing all the useful libraries
```

```python
dataset = pd.read_csv("AsiaCovidCases.csv")#reading the file
```

```python
dataset#previewing the dataset to train
```

```python
dataset.drop("TotalTestsPerMillion", axis=1, inplace=True)#removing the non useful coloumns using inplace to make the permanent changes to the dataset 
```

```python
dataset.drop("TotalDeathsPerMillion", axis=1, inplace=True)# doing the same thing dropping the couumn
```

```python
dataset.drop("TotalTests", axis=1, inplace=True)
```

```python
dataset.drop("TotalCasesPerMillion", axis=1, inplace=True)
```

```python
dataset.drop("ActiveCases", axis=1, inplace=True)
```

```python
Country = pd.get_dummies(dataset["Country"])
```

```python
Country = Country.astype(bool)
```

```python
dataset = dataset.drop("Country", axis=1)
```

```python
dataset = pd.concat([dataset, Country], axis=1)
```

```python
dataset.drop("ID", axis=1, inplace=True)
```

```python
dataset#viewing the dataset before training
```

```python
X = dataset.drop("TotalDeaths", axis=1)#alocating the x column to ignore any irrelavant error
```

```python
Y = dataset["TotalDeaths"]#alocating the target coloumn manually
```

```python
X_train, X_test, y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=9) #training the model
```

```python
#algorithm1
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators=40, min_samples_leaf=2)# the functions are taken from scikit learn the parameters are set on its highest performance capacity after playing around various limit.
regressor.fit(X_train, y_train)#training the model
Y_pred = regressor.predict(X_test)
predictions = regressor.predict(X_test)
print(predictions)#printing the predictions
r_square = metrics.r2_score(Y_test, predictions)#cheching the accuracy aswell
print(r_square)
```

```python
Y_test #just to check that it is following the command in which Y has been given properly
```

```python
# Ploting the randomforestregression graph
plt.scatter(Y_test, Y_pred, alpha=0.7)
plt.plot([min(Y_test), max(Y_test)], [min(Y_test), max(Y_test)], 'g--', label="Perfect Prediction")
plt.xlabel("Observed Values")
plt.ylabel("Model predition values")
plt.title("(Random Forest Regression Graph)")
plt.legend()
plt.show()
```

```python
#algorith 2
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(criterion='squared_error', max_depth=10)#the criterion function is the extra function taken from scikit learn on both squared_error and friedman_mse giving the alost same performance but lower at the absolute error and the other two read in the lecture.
regressor.fit(X_train, y_train)
Y_pred = regressor.predict(X_test)
predictions = regressor.predict(X_test)
print(predictions)
r_square = metrics.r2_score(Y_test, predictions)
print(r_square)
```

```python
#plotting the graph for dtr
plt.scatter(Y_test, Y_pred, alpha=0.8, label="Predicted Values")
plt.plot([min(Y_test), max(Y_test)], [min(Y_test), max(Y_test)], 'r--', label="Perfect Prediction Line")
plt.xlabel("Observed Values")
plt.ylabel("Model predition values")
plt.title("(Decision Tree Regression Graph)")
plt.legend()
plt.show()
```
